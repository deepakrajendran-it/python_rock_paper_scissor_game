##### Add Game Logic

```python
print("=== Rock, Paper, Scissors Game ===")

choices = ["rock", "paper", "scissors"]

user_choices = input("Enter your choice 'Rock, Paper , Scissors'  - ").strip().lower()

if user_choices not in choices:
    print("Invalid choice, Please enter Rock, Paper, Scissors as input")

computer_choice = random.choice(choices)

print("User ",user_choices)
print("Computer  ",computer_choice)

if user_choices == computer_choice:
    print("Tie")
elif (
    (user_choices  == "rock" and computer_choice == "scissors") or
    (user_choices  == "paper" and computer_choice == "rock") or
    (user_choices  == "scissors" and computer_choice == "paper")
    ):
    print("You Win")
else:
    print("Computer Win")
```

```python
print("=== Rock, Paper, Scissors Game ===")

choices = ["rock", "paper", "scissors"]

output=[]

while True:
    user_choices = input("Enter your choice 'Rock, Paper , Scissors'  - ").strip().lower()

    if user_choices not in choices:
        print("Invalid choice, Please enter Rock, Paper, Scissors as input")

    computer_choice = random.choice(choices)

    if user_choices == computer_choice:
        result = "Tie"
    elif (
        (user_choices  == "rock" and computer_choice == "scissors") or
        (user_choices  == "paper" and computer_choice == "rock") or
        (user_choices  == "scissors" and computer_choice == "paper")
        ):
        result = "You Win"
        
    else:
        result = "Computer Win"
        
    output.append({
    "User Input": user_choices,
    "Computer Input": computer_choice,
    "Result": result })
    
    print("-" * 60)
    print(f"{'User Input':<15}{'Computer Input':<20}{'Result':<20}")
    print("-" * 60)
    
    for game in output:
        print(f"{game['User Input']:<15}{game['Computer Input']:<20}{game['Result']:<20}")
        
    print("-" * 60)
    play_again = input("Play Again (yes/no): ").strip().lower()
    
    if play_again!="yes":
        print("Game Over")
        break 
```

```python
print("=== Rock, Paper, Scissors Game ===")

choices = ["r", "p", "s"]

output=[]

user_count =0;
computer_count=0;
tie_count =0;

while True:
    user_choices = input("Enter your choice 'Rock - r, Paper -p , Scissors - s'  - ").strip().lower()

    if user_choices not in choices:
        print("Invalid choice, Please enter Rock, Paper, Scissors as input")

    computer_choice = random.choice(choices)

    if user_choices == computer_choice:
        result = "Tie"
        tie_count+=1
    elif (
        (user_choices  == "r" and computer_choice == "s") or
        (user_choices  == "p" and computer_choice == "r") or
        (user_choices  == "s" and computer_choice == "p")
        ):
        result = "You Win"
        user_count +=1
    else:
        result = "Computer Win"
        computer_count+=1
        
    output.append({
    "User Input": user_choices,
    "Computer Input": computer_choice,
    "Result": result })
    
    print("-" * 60)
    print(f"{'Aashmi Input':<15}{'Computer Input':<20}{'Result':<20}")
    print("-" * 60)
    
    
    for game in output:
        print(f"{game['User Input']:<15}{game['Computer Input']:<20}{game['Result']:<20}")
        
    print("-" * 60)
    
    
    print(f"{'Aashmi':<15}{'Computer':<20}{'Tie':<20}")
    print(f"{user_count:<15}{computer_count:<20}{tie_count:<20}")
    print("-" * 60)
     
    play_again = input("Play Again (yes -y/no -n): ").strip().lower()
    
    if play_again=="yy":
        print("Game Over")
        break 
```
